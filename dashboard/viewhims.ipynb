```python
import sqlite3
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

```python
conn=sqlite3.connect('hims.db')
```

```python
cur=conn.cursor()
cur.execute("""
UPDATE patients
SET fullname = ?,
    age = ?,
    gender = ?,
    email = ?
WHERE rowid = (SELECT rowid FROM patients ORDER BY rowid LIMIT 1)
""", ('Sana Bhatia', 20, 'Female', 'sana@gmail.com'))
conn.commit()
```

```python
df=pd.read_sql_query('select* from patients ;',conn)
df
```

```python
total_patients=df['patient_id'].nunique()
total_patients
```

```python
Min_age=df['age'].min()
Max_age=df['age'].max()
Avg_age=df['age'].mean()
```

```python
Gdf=pd.DataFrame(df['gender'].value_counts()).reset_index()
```

```python
Gdf
```

```python
fig=px.bar(Gdf,color='gender')
fig.show()
```

```python
GMDF=df.groupby(['gender','marital_status']).agg({'patient_id' : 'nunique'}).reset_index()
```

```python
GMDF
```

```python
px.bar(GMDF,x='gender',y='patient_id',color='marital_status')
```

```python
px.bar(df.groupby('has_insurance').agg({'patient_id' : 'nunique'}))
```
